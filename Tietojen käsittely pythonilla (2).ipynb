```python
taulukko = pd.crosstab(df['gender'], df['sat_salary'])
print("Ristiintaulukko (Sukupuoli ja palkkatyytyväisyys):")
print(taulukko)
```

```python
taulukko.plot(kind='bar')
plt.title('Palkkatyytyväisyys ja sukupuoli (1=Mies, 2=Nainen)')
plt.xlabel('Sukupuoli')
plt.show()
```
