```python
# Open the data
df = pd.read_excel('WorkPlaceSatisfactionSurveyData.xlsx')
```

```python
df.columns = ['Number', 'gender', 'age', 'family', 'education', 'years_of_service', 'salary', 'sat_management', 'sat_colleques', 'sat_workingEnvironment', 'sat_salary', 'sat_tasks', 'holidayCabin', 'gym', 'muscleCare', 'healtcare',]
```

## Part: Create some classification for quantitative variable

```python
min(df['age'])
```

```python
max(df['age'])
```

```python
bins = [500, 1000, 2000, 3000, 4000, 7000]
```

```python
#df['age'] = pd.cut(df['age'], bins = bins)
```

```python
df
```

```python
df3 = pd.crosstab(df['age'], 'Count')

df3.columns.name = ''

# Percentages
n = df3['Count'].sum()
df3['%'] = df3 / n * 100
```

```python
df3
```

```python
type(df3.index.values[0])
```

```python
# Note, that income column has now values of type Interval!
# We would like to add row "Total" into the index, but the Index can contain only one type of values
# for that reason we need to change the type of index elements to be string!

df3.index = df3.index.astype(str)
df3.loc['Total'] = df3.sum()
df3
```

```python
type(df3.index.values[0])
```

```python
# Finalizing the layout
df3.style.format({'Count': '{:.0f}', '%': '{:.1f} %'})
```

## Part: Also create Histograms using the counts and using percentages

 - let's cover the counts part first

```python
df['age'].hist(bins = bins)

plt.xlabel('age')
plt.ylabel('Count, n=' + str(n))
```

- Then let's cover percentages

```python
df['age'].hist(bins = bins, weights = np.ones(n) / n * 100)

plt.xlabel('age')
plt.ylabel('%, n=' + str(n))
```
